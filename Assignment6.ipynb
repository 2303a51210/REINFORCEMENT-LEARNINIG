```python


import gymnasium as gym


def random_agent(env_name, episodes=10):

    # Create environment
    env = gym.make(env_name)

    episode_rewards = []

    print("\n======================================")
    print("Environment:", env_name)
    print("======================================")

    # Repeat for multiple episodes
    for episode in range(1, episodes + 1):

        # S0 = Reset(E)
        state, info = env.reset()

        G = 0          # Total reward
        T = 0          # Number of steps

        terminated = False
        truncated = False

        # print("\nEpisode:", episode)
        # print("Initial State:", state)

        # Continue until episode terminates
        while not (terminated or truncated):

            # At ~ Uniform(A)
            action = env.action_space.sample()

            # (S(t+1), R(t)) = Step(S(t), A(t))
            next_state, reward, terminated, truncated, info = env.step(action)

            # G = G + R(t)
            G = G + reward

            # T = T + 1
            T = T + 1

            # print(
            #     "State:", state,
            #     "Action:", action,
            #     "Reward:", reward,
            #     "Next State:", next_state
            # )

            # Move to next state
            state = next_state

        # Average reward = G / T
        average_reward = G / T

        episode_rewards.append(G)

        print("\nEpisode:", episode)
        print("Total Reward (G):", G)
        print("Number of Steps (T):", T)
        print("Average Reward (G/T):", round(average_reward, 2))

    # Average reward over multiple episodes
    G_avg = sum(episode_rewards) / episodes

    print("\n======================================")
    print("FINAL RESULTS")
    print("======================================")

    for i in range(episodes):
        print(
            "Episode", i + 1,
            "-> Total Reward =", episode_rewards[i]
        )

    print("\nAverage Total Reward over all episodes:\n", round(G_avg, 2))

    env.close()


# -----------------------------------
# 1. CliffWalking
# -----------------------------------

random_agent("CliffWalking-v1", 10)


# -----------------------------------
# 2. Taxi
# -----------------------------------

random_agent("Taxi-v4", 10)

```
